# Visualizing example audio file

```python
random_file_name1 = "rock.00026.wav"
y, sr = librosa.load(random_file_name1, sr=44100)
```

### What is returned?
- Audio time series: `y` -> amplitude over time
- Sampling rate of y: `sr` -> how many samples over 1 second

```python
y
```

```python
print(f'''Shape of audio time series: {y.shape}
Value of sample rate: {sr}''')
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveshow(y, sr=sr)
```

## Playing sound

```python
Audio(data=y, rate=sr)
```

# Visualizing melspectogram

```python
def plot_melspectogram(y, sr):
    spectogram = librosa.feature.melspectrogram(y=y, sr=sr)
    spectogram_db = librosa.power_to_db(spectogram, ref=np.max)
    
    plt.figure(figsize=(10, 4))
    librosa.display.specshow(spectogram_db, sr=sr, x_axis='time', y_axis='mel', cmap='inferno')
    plt.colorbar(format='%2.0f dB')
    plt.title('Melspectogram')
    plt.tight_layout()
    plt.show()

plot_melspectogram(y, sr)
```
